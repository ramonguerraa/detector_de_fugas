## 2. GENERACIÓN DE COMPRAS (Evolución de Costos e Inventario).

* Simularemos compras a proveedores a lo largo de 6 meses (Enero - Junio).

```python
fecha_inicio = datetime(2026, 1, 1)
registros_compras = []

for prod_id, info in productos_base.items():
    costo_actual = info['costo_inicial']
    for mes in range(6):
        fecha_compra = fecha_inicio + timedelta(days=mes*30 + np.random.randint(1, 10))
        
        # Sembramos la PRIMERA FUGA: Inflación de costos no indexada
        if prod_id in ['PROD002', 'PROD005', 'PROD007', 'PROD011', 'PROD016', 'PROD023', 'PROD025']:
            costo_actual *= np.random.uniform(1.08, 1.15) # Aumento del 8% al 15% mensual
        else:
            costo_actual *= np.random.uniform(1.01, 1.04) # Aumento normal del 1% al 4%
            
        cantidad = int(np.random.choice([100, 200, 500, 1000] if info['categoria'] != 'Lubricantes' else [250, 500, 1000]))
        
        registros_compras.append({
            'fecha': fecha_compra.strftime('%Y-%m-%d'),
            'producto_id': prod_id,
            'descripcion': info['descripcion'],
            'categoria': info['categoria'],
            'marca': info['marca'],
            'cantidad_comprada': cantidad,
            'costo_unitario': round(costo_actual, 2)
        })

df_compras = pd.DataFrame(registros_compras)
```

## 3. GENERACIÓN DE VENTAS (Comportamiento Comercial).

* Simularemos transacciones diarias de ventas.

```python
registros_ventas = []
fecha_actual = fecha_inicio

while fecha_actual <= datetime(2026, 6, 15):
    # Número de ventas por día
    num_ventas = np.random.randint(35, 50)
    for _ in range(num_ventas):
        prod_id = np.random.choice(list(productos_base.keys()))
        info = productos_base[prod_id]
        
        # Encontrar el costo vigente de ese producto en esa fecha
        costo_vigente = df_compras[(df_compras['producto_id'] == prod_id) & 
                                   (df_compras['fecha'] <= fecha_actual.strftime('%Y-%m-%d'))]['costo_unitario'].last_valid_index()
        
        costo_v = df_compras.loc[costo_vigente, 'costo_unitario'] if costo_vigente is not None else info['costo_inicial']
        
        # Definición estándar de precio (Costo Inicial + 40% de margen)
        precio_venta_estandar = info['costo_inicial'] * 1.40
        
        # Sembramos la SEGUNDA FUGA: Precio congelado vs Costo dinámico
        # Si es un producto de la lista, el vendedor sigue usando el precio viejo basado en el costo inicial, ignorando las alzas.
        if prod_id in ['PROD001', 'PROD002', 'PROD007', 'PROD016', 'PROD020']:
            precio_final = precio_venta_estandar # No sube jamás
        else:
            precio_final = costo_v * np.random.uniform(1.30, 1.45) # Margen sano sobre costo actual
            
        # Sembramos la TERCERA FUGA: Descuentos "Hormiga" u Operaciones Sospechosas
        # Ciertos días (ej. fines de semana), el producto PROD001 se vende con un descuento excesivo por un vendedor X
        vendedor = np.random.choice(['Vendedor_A', 'Vendedor_B', 'Vendedor_C'], p=[0.5, 0.3, 0.2])
        if vendedor == 'Vendedor_C' and prod_id in ['PROD001', 'PROD025'] and np.random.rand() > 0.4:
            precio_final = costo_v * 0.90 # ¡Vendiendo por debajo del costo!
            
        cantidad_vendida = np.random.randint(1, 6) if info['categoria'] != 'Lubricantes' else np.random.randint(2, 24)
        
        registros_ventas.append({
            'factura': f'FAC-{len(registros_ventas)+1000}',
            'fecha': fecha_actual.strftime('%Y-%m-%d'),
            'vendedor': vendedor,
            'producto_id': prod_id,
            'cantidad_vendida': cantidad_vendida,
            'precio_unitario': round(precio_final, 2),
            'costo_asociado_simulado': round(costo_v, 2) # Guardamos esto para verificar luego
        })
        
    fecha_actual += timedelta(days=1)

df_ventas = pd.DataFrame(registros_ventas)
```

## 4. EXPORTAR A CSV.

```python
df_compras.to_csv('data/compras_proveedores.csv', index=False)
df_ventas.to_csv('data/ventas_diarias.csv', index=False)

print(f"¡Data de laboratorio ampliada generada con éxito!")
print(f"Registros de compras: {len(df_compras)} | Registros de ventas: {len(df_ventas)}")
```
